#Example 7:
Create two vectors in 2-dimensional (2D) space and plot them using Matplotlib.

Solution: A vector in 2D space can created as a 1D ndarray with two elements as:

V = np.array([1,1])

W = np.array([-2.0,-1])

To plot the above vectors, we will use the quiver() function from the pyplot module of Matplotlib
library. The quiver() function is used to create 2D field or velocity plots and is perfect for our purpose
of vector plotting. Here is an example of plotting the above vectors.

```python
import numpy as np
import matplotlib.pyplot as plt
""" Problem 2: Create 2-dimensional position vectors and plot them. """
origin = np.array([0,0])
V = np.array([1,1])
W = np.array([-2.0,-1])
fig, ax = plt.subplots()
# Add vectors V and W to the plot
ax.quiver(origin[0], origin[1], V[0], V[1], angles='xy', scale_units='xy', scale=1, color='r')
ax.quiver(origin[0], origin[1], W[0], W[1], angles='xy', scale_units='xy', scale=1, color='b')
ax.text(origin[0]-0.1, origin[1]+0.1, "O")
ax.text(V[0], V[1], "v")
ax.text(W[0]-0.125, W[1]-0.1, "w")
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.grid()
plt.show()
```

#Task: 6
a. Create three vectors in 3D space and plot them using Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt

# Enable 3D plotting
from mpl_toolkits.mplot3d import Axes3D

origin = np.array([0,0,0])
U= np.array([1, 0.25 ,0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot vectors
# quiver for 3D does not use 'angles' or 'scale_units'
ax.quiver(origin[0], origin[1], origin[2], U[0], U[1], U[2], color='r', arrow_length_ratio=0.1, label='U')
ax.quiver(origin[0], origin[1], origin[2], V[0], V[1], V[2], color='b', arrow_length_ratio=0.1, label='V')
ax.quiver(origin[0], origin[1], origin[2], W[0], W[1], W[2], color='g', arrow_length_ratio=0.1, label='W')

# Set plot limits and labels
max_coord = max(np.max(np.abs(U)), np.max(np.abs(V)), np.max(np.abs(W))) * 1.5
ax.set_xlim([-max_coord, max_coord])
ax.set_ylim([-max_coord, max_coord])
ax.set_zlim([-max_coord, max_coord])
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('Three Vectors in 3D Space')
ax.legend()
plt.grid(True)
```

b. Create two vectors in 2D space, and add them to create a third vector, and plot all three vectors
using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create two vectors
U = np.array([-0.5, 0.5])
V = np.array([1,1])
#The two vectors are added as
W = U + V # adding two vector
fig, ax = plt.subplots()
ax.quiver(origin[0], origin[1], U[0], U[1], angles='xy', scale_units='xy', scale=1, color='r')
ax.quiver(origin[0], origin[1], V[0], V[1], angles='xy', scale_units='xy', scale=1, color='b')
ax.quiver(origin[0], origin[1], W[0], W[1], angles='xy', scale_units='xy', scale=1, color='g')
ax.text(0.125, 0, "O")
ax.text(W[0], W[1], "U+V=W")
ax.text(U[0], U[1], "U")
ax.text(V[0], V[1], "V")
# Axes settings
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.grid()

plt.show()
```

d. Write a function to calculate the direction cosines of the vector. For example, let the vector be
O = (1.5, 1.5, 2.0).

```python
import numpy as np
def direction_cosines(vector):
    magnitude = np.linalg.norm(vector)
    direction_cosines = vector / magnitude
    return direction_cosines
O=np.array([1.5,1.5,2.0])
print("Direction cosines = ",np.round(direction_cosines(O),2))
```

e. Write a function to calculate Scalar (dot) product of two vectors. For example, let the vectors
be a= (1.5, 1.5, 2.0) and b = (−1.5, 0.5, 2.0).

```python
import numpy as np
def scalar_product(a, b):
    return np.dot(a, b)
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
print("Scalar product = ",scalar_product(a,b))
```

f. Write a function to calculate Vector (cross) product of two vectors. For example, let the vectors
be a = (1.5, 1.5, 2.0) and b= (−1.5, 0.5, 2.0).

```python
import numpy as np
def cross_product(a, b):
    return np.cross(a, b)
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
print("Cross Product:", cross_product(a, b))
```
