# **Scatter**
Scatter graphs will plot one point per observation, we can use these to graph independent
variables to see if there is a correlation. Python will require two sets of variables for a scatter
graph, one for the x-axis and another for the corresponding y axis. These sets will need to be of
the same length and work best as lists, array, or array-like data types. We can call ‘pyplot’ to plot
a scatter with the pyplot.scatter(x,y) call for example.

Example 1:

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8,9]
y=[3,4,5,6,7,8,9,10,11]
#plt.plot(x,y)
plt.scatter(x,y,color='red',marker='^')
plt.show()
```

Example 2:

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8,9]
y=[3,4,5,6,7,1,9,2,5]
#plt.plot(x,y)
plt.scatter(x,y,color='red',marker='^')
x1=[1,5,6,8,12]
y1=[2,6,7,11,13]
plt.scatter(x1,y1,color='blue',marker='s')
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.title("Scatter plot")
plt.show()
```

Task 2:
Following data shows the change in population of different animals from 2017 to 2022.
a. Create a pie chart for the population of each year
b. Can we show all the pie chart on the single pane? Write the code.

```python
import matplotlib.pyplot as plt
y=[100,205,307,409,200]
years=[2017,2018,2019,2020,2021,2022]
animals=['Bears','Dolphins','Whales']
y=[[8,150,80],[54,77,54],[93,32,100],[116,11,76],[137,6,93],[184,1,72]]
plt.title("Population of animals")
plt.xticks([])
plt.yticks([])
for i in range(6):
  plt.subplot(2,3,i+1)
  plt.title(str(years[i]))
  plt.pie(y[i],labels=animals)
plt.show()
```

C- plt.legend(title = "Car Colours:")

```python
import matplotlib.pyplot as plt
x=['Red','Silver','Blue','Green']
y=[123,230,78,12]
myexplode = [0.3, 0, 0, 0]
plt.legend(title="Car Colours:")
plt.pie(y,labels=x,colors=x,explode = myexplode,shadow = True)
```
