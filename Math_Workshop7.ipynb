Part 1: Create and Understand Linear Systems:

Example 1: We will start with a simple problem. Consider the following linear systems:

a.

x+3y=5

x-2y=-5

x+y=3

```python
import numpy as np
import matplotlib.pyplot as plt
""" Part 1 - Problem 1: Solve linear systems graphically. """
x = np.linspace(-5,5,100)
y1 = (5-x)/3
y2 = (5+x)/2
y3 = 3-x
fig, ax = plt.subplots()
ax.plot(x,y1, color='r')
ax.plot(x,y2, color='g')
ax.plot(x,y3, color='b')
ax.text(0.5,0.6, '$x+3y = 5$', color='r')
ax.text(-3,0.5,'$x-2y = -5$', color='g')
ax.text(1.2,2, '$x+y = 3$', color='b')
ax.set_xlim(-4,4)
ax.set_ylim(-2,6)
ax.axvline(color='k',linewidth = 1)
ax.axhline(color='k',linewidth = 1)
ax.set_xticks(list(range(-4,5)))
ax.set_aspect('equal')
ax.grid(True,ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Conclusion: All the lines have unique solutions.

#**Numpy.linalg**

linalg is a submodule in NumPy, a popular Python library used for numerical computations, particularly in the domain of linear algebra. linalg stands for "linear algebra". This submodule provides functions and classes for various linear algebra operations, such as matrix decompositions, eigenvalue computations, matrix multiplication, solving linear equations, and many others.

```python
import numpy as np
a=np.array([[1,2],[3,4]])
a_inv=np.linalg.inv(a)
print(a_inv)
```

Consider the following 10 points in X-Y plane: (−3, −7.6), (−2,1), (−1, 4.8), (0, 5), (1, 2.8), (2, −0.6), (3, −4), (4, −6.2), (5, −6), (7, 6.4). Plot these points in X-Y plane. Connect the points to guess the polynomial. Once you guess your polynomial, formulate a Linear System and then solve it to find the coefficient of the polynomial. Finally, check if the polynomial fits along with the above given points.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])
plt.plot(x, y, color='blue', label='Data Points',marker='o',markerfacecolor='red',markeredgecolor='red')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()
```

We see that the points form an 'S' shape with 2 curves, so the polynomial is of 3rd degree

```python
x = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])

# Create the Vandermonde matrix
A = np.vander(x[:4], 4)
print(A)
# Solve the system Ax = b
coeffs = np.linalg.solve(A,y[:4])
print(coeffs)
a,b,c,d=coeffs
print(f"Coefficients: {coeffs}")
y_check = a*x**3 + b*x**2 + c*x + d
matches = np.allclose(y, y_check)
print(f"Perfect fit for all points? {matches}")
```
